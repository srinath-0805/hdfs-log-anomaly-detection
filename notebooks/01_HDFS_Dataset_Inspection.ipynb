```python
# Calculate event sequence lengths
import re

# Read only the Features column
features_df = con.execute("""
    SELECT Features
    FROM event_traces
""").df()

# Count event IDs in each sequence
features_df["sequence_length"] = (
    features_df["Features"]
    .apply(lambda x: len(re.findall(r"E\d+", x)))
)

# Statistical summary
display(
    features_df["sequence_length"].describe()
)
```

```python
# Find the sequence-length percentiles
percentiles = features_df["sequence_length"].quantile(
    [0.90, 0.95, 0.99, 0.995]
)

print("Sequence Length Percentiles:")
display(percentiles)

print("\nSequences exceeding selected lengths:")

for limit in [20, 25, 30, 40, 50, 100]:
    count = (features_df["sequence_length"] > limit).sum()

    percentage = count / len(features_df) * 100

    print(
        f"Above {limit}: {count:,} sequences "
        f"({percentage:.2f}%)"
    )
```

```python
# Validate long sequences against anomaly labels
long_sequence_analysis = con.execute("""
    WITH sequence_data AS (
        SELECT
            e.BlockId,
            array_length(
                string_split(e.Features, ',')
            ) AS sequence_length,
            a.Label AS actual_label

        FROM event_traces e

        INNER JOIN anomaly_labels a
            ON e.BlockId = a.BlockId
    )

    SELECT
        CASE
            WHEN sequence_length > 40
                THEN 'Above 40'
            ELSE '40 or fewer'
        END AS sequence_group,

        COUNT(*) AS total_blocks,

        SUM(
            CASE WHEN actual_label = 'Anomaly'
                 THEN 1 ELSE 0 END
        ) AS anomaly_blocks,

        ROUND(
            100.0 * SUM(
                CASE WHEN actual_label = 'Anomaly'
                     THEN 1 ELSE 0 END
            ) / COUNT(*),
            2
        ) AS anomaly_percentage

    FROM sequence_data

    GROUP BY sequence_group
    ORDER BY sequence_group
""").df()

display(long_sequence_analysis)
```

```python
# Analyze sequence length as an anomaly indicator
sequence_risk = con.execute("""
    WITH sequence_data AS (
        SELECT
            e.BlockId,
            array_length(
                string_split(e.Features, ',')
            ) AS sequence_length,
            a.Label AS actual_label
        FROM event_traces e
        INNER JOIN anomaly_labels a
            ON e.BlockId = a.BlockId
    )

    SELECT
        CASE
            WHEN sequence_length <= 15 THEN '2-15'
            WHEN sequence_length <= 20 THEN '16-20'
            WHEN sequence_length <= 25 THEN '21-25'
            WHEN sequence_length <= 30 THEN '26-30'
            WHEN sequence_length <= 40 THEN '31-40'
            ELSE 'Above 40'
        END AS sequence_range,

        COUNT(*) AS total_blocks,

        SUM(
            CASE WHEN actual_label = 'Anomaly'
            THEN 1 ELSE 0 END
        ) AS anomalies,

        ROUND(
            100.0 * SUM(
                CASE WHEN actual_label = 'Anomaly'
                THEN 1 ELSE 0 END
            ) / COUNT(*), 2
        ) AS anomaly_rate

    FROM sequence_data
    GROUP BY sequence_range
    ORDER BY
        CASE sequence_range
            WHEN '2-15' THEN 1
            WHEN '16-20' THEN 2
            WHEN '21-25' THEN 3
            WHEN '26-30' THEN 4
            WHEN '31-40' THEN 5
            ELSE 6
        END
""").df()

display(sequence_risk)
```

```python
# Validate event IDs
import re
from collections import Counter

# Count event IDs across all sequences
event_counter = Counter()

for sequence in features_df["Features"]:
    event_ids = re.findall(r"E\d+", sequence)
    event_counter.update(event_ids)

# Display event frequencies
event_frequency = pd.DataFrame(
    event_counter.items(),
    columns=["EventId", "Frequency"]
)

event_frequency["EventNumber"] = (
    event_frequency["EventId"]
    .str.extract(r"(\d+)")
    .astype(int)
)

event_frequency = event_frequency.sort_values(
    "EventNumber"
).drop(columns="EventNumber")

display(event_frequency)
```

```python
print("Total unique event IDs:", len(event_counter))

print(
    "Event IDs:",
    sorted(
        event_counter.keys(),
        key=lambda x: int(x[1:])
    )
)
```

```python
# Create event-to-number mapping
# Create numerical mapping for all event IDs

event_vocabulary = sorted(
    event_counter.keys(),
    key=lambda x: int(x[1:])
)

event_to_id = {
    event: index + 1
    for index, event in enumerate(event_vocabulary)
}

id_to_event = {
    value: key
    for key, value in event_to_id.items()
}

print("Total event types:", len(event_to_id))
print("\nEvent mapping:")
print(event_to_id)
```

```python
# Convert event sequences into numerical sequences
# Convert five sample sequences first
import re

# Take five sample event sequences
sample_sequences = features_df["Features"].head(5)

# Convert event IDs into numerical IDs
encoded_samples = []

for sequence in sample_sequences:

    events = re.findall(r"E\d+", sequence)

    encoded_sequence = [
        event_to_id[event]
        for event in events
    ]

    encoded_samples.append(encoded_sequence)


# Display original and encoded sequences
for i in range(5):

    print(f"\nSequence {i + 1}")

    print("Original:")
    print(sample_sequences.iloc[i])

    print("\nEncoded:")
    print(encoded_samples[i])
```

```python
# Convert all event sequences
import csv
import re

from pathlib import Path

# Output location
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = PROCESSED_DIR / "encoded_sequences.csv"

# Read event sequences and their actual ground-truth labels
query = """
    SELECT
        e.BlockId,
        e.Features,
        a.Label AS ActualLabel
    FROM event_traces e
    INNER JOIN anomaly_labels a
        ON e.BlockId = a.BlockId
"""

cursor = con.execute(query)

total_processed = 0
batch_size = 10000

# Stream records in batches
with open(OUTPUT_FILE, "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow([
        "BlockId",
        "EncodedFeatures",
        "SequenceLength",
        "ActualLabel"
    ])

    while True:

        batch = cursor.fetch_df_chunk(batch_size)

        if batch.empty:
            break

        for row in batch.itertuples(index=False):

            events = re.findall(r"E\d+", row.Features)

            encoded_sequence = [
                event_to_id[event]
                for event in events
            ]

            writer.writerow([
                row.BlockId,
                ",".join(map(str, encoded_sequence)),
                len(encoded_sequence),
                row.ActualLabel
            ])

        total_processed += len(batch)

        print(
            f"Processed: {total_processed:,} / 575,061"
        )

print("\nConversion completed!")
print("Total sequences:", total_processed)
print("Saved location:", OUTPUT_FILE)
```

```python
# Preview the split distribution
split_preview = con.execute(f"""
    WITH assigned_data AS (
        SELECT
            BlockId,
            ActualLabel,
            HASH(BlockId) % 100 AS split_value
        FROM read_csv_auto('{OUTPUT_FILE.as_posix()}')
    )
    SELECT
        CASE
            WHEN ActualLabel = 'Anomaly' THEN 'Test'
            WHEN split_value < 70 THEN 'Train'
            WHEN split_value < 85 THEN 'Validation'
            ELSE 'Test'
        END AS Dataset,
        ActualLabel,
        COUNT(*) AS Total
    FROM assigned_data
    GROUP BY Dataset, ActualLabel
    ORDER BY Dataset, ActualLabel
""").df()

display(split_preview)
```

```python
# Save the split assignments
# Now we'll save the split information so we can reuse exactly the same assignment during model training and evaluation.
SPLIT_FILE = OUTPUT_FILE.parent / "split_assignments.csv"

con.execute(f"""
    COPY (
        SELECT
            BlockId,
            ActualLabel,
            CASE
                WHEN ActualLabel = 'Anomaly' THEN 'Test'
                WHEN HASH(BlockId) % 100 < 70 THEN 'Train'
                WHEN HASH(BlockId) % 100 < 85 THEN 'Validation'
                ELSE 'Test'
            END AS Dataset
        FROM read_csv_auto('{OUTPUT_FILE.as_posix()}')
    )
    TO '{SPLIT_FILE.as_posix()}'
    (HEADER, DELIMITER ',')
""")

print("Split assignments saved successfully!")
print("Location:", SPLIT_FILE)
```

```python
# Verify the saved split file
saved_split = con.execute(f"""
    SELECT
        Dataset,
        ActualLabel,
        COUNT(*) AS Total
    FROM read_csv_auto('{SPLIT_FILE.as_posix()}')
    GROUP BY Dataset, ActualLabel
    ORDER BY Dataset, ActualLabel
""").df()

display(saved_split)

print(
    "Total records:",
    saved_split["Total"].sum()
)
```
